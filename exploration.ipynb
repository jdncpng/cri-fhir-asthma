```python
import json
from collections import Counter
```

```python
# loading the file from data folder
patient_files = "data/Dyan186_Purdy2_61b11ebf-b7c1-0bb8-2ded-c5405bd62015.json"

with open(patient_files, "r") as f:
    data = json.load(f)
```

```python
# raw structure

# top-level keys
print(data.keys())
# number of entries
print(len(data["entry"]))
```

```python
# showing all unique top-level resources
resource_types = {entry["resource"]["resourceType"] for entry in data["entry"]}

for rt in sorted(resource_types):
    print(rt)
```

```python
# 1.2.
patient = next(
    entry["resource"] for entry in data["entry"]
    if entry["resource"]["resourceType"] == "Patient"
)

print(json.dumps(patient, indent=2))
```

```python
# extracting name and birthdate
name = patient["name"][0]
full_name = f"{name.get('given', [''])[0]} {name.get('family')}"
birth_date = patient["birthDate"]

print(f"Full Name: {full_name}")
print(f"Date of Birth: {birth_date}")
```
